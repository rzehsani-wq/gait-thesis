```python
# Cell 1: Setup
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../runs_master.csv")
print(f"Loaded {len(df)} runs")
df.head()
```

```python
# Cell 2: Filter to a comparison of interest
# Example: compare nm→cl across SSL pretraining sources
mask = (df["region"] == "R1") & (df["probe_cond"] == "cl")
view = df[mask][["run_name", "ckpt", "gallery_per_subject", "best.top1", "best.top5"]]
view.sort_values("best.top1", ascending=False)
```

```python
# Cell 3: A simple plot — Rank-1 by SSL initialization
plt.figure(figsize=(8, 5))
sns.barplot(data=view, x="ckpt", y="best.top1")
plt.title("nm→cl Rank-1 by SSL initialization (R1)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("../figures/nm2cl_by_ssl_init.png", dpi=150)
plt.show()
```
